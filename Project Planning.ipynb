## Project Planning ##

To help with project planning a Gantt Chart was created to help stay on task

```python
# google gemini was consulted with the code for the chart
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

# 1. Define tasks and their corresponding start and end dates based on course calendar
tasks = [
    "Dataset Setup & EDA",
    "Baseline ML Model Prep",
    "Midterm Exam (Milestone)",
    "Advanced Modeling & XGBoost",
    "Kaggle Submissions & Tuning",
    "Project Progress Check (Milestone)",
    "Model Interpretation & Writing",
    "Final Polish & Slides",
    "Project Deadline & Presentation"
]

# Convert date strings to datetime objects for 2026
start_dates = [
    datetime(2026, 8, 31),  # Class begin
    datetime(2026, 9, 28),  
    datetime(2026, 10, 21), # Midterm exam
    datetime(2026, 10, 22), 
    datetime(2026, 11, 2),  
    datetime(2026, 11, 20), # Progress check
    datetime(2026, 11, 21), 
    datetime(2026, 12, 1),  
    datetime(2026, 12, 11)  # Project deadline
]

end_dates = [
    datetime(2026, 9, 27),
    datetime(2026, 10, 20),
    datetime(2026, 10, 21), # Single day milestone
    datetime(2026, 11, 1),
    datetime(2026, 11, 19),
    datetime(2026, 11, 20), # Single day milestone
    datetime(2026, 11, 30),
    datetime(2026, 12, 10),
    datetime(2026, 12, 11)  # Single day milestone
]

# 2. Calculate durations in days
durations = [(end - start).days for start, end in zip(start_dates, end_dates)]
# Ensure milestones have a small visible width (at least 1 day)
durations = [d if d > 0 else 1 for d in durations]
start_nums = mdates.date2num(start_dates)

# 3. Create the plot
fig, ax = plt.subplots(figsize=(11, 6))

# Use distinct colors for regular tasks vs. course milestones
bar_colors = ['#4C72B0', '#4C72B0', '#DD8452', '#4C72B0', '#4C72B0', '#DD8452', '#4C72B0', '#4C72B0', '#C44E52']

ax.barh(range(len(tasks)), durations, left=start_nums, align='center', 
        color=bar_colors, edgecolor='#2C3E50', alpha=0.85, height=0.6)

# Format y-axis labels
ax.set_yticks(range(len(tasks)))
ax.set_yticklabels(tasks, fontsize=10, fontweight='bold')
ax.invert_yaxis()  # Top-down order

# Format x-axis dates (spanning from August to December)
ax.xaxis_date()
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))
ax.xaxis.set_major_locator(mdates.MonthLocator())
plt.xticks(rotation=0)

# Add labels, title, and grid
plt.xlabel('Date (2026)', fontsize=11, fontweight='bold')
plt.title('STT 481 Capstone Project Timeline & Course Milestones', fontsize=14, fontweight='bold', pad=15)
plt.grid(axis='x', linestyle='--', alpha=0.6)

plt.tight_layout()

# Save and show chart
plt.savefig('stt481_official_gantt.png', dpi=300)
plt.show()
```
