### LASSO

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
```

```python
y = clean_train["SalePrice"]
X = clean_train.drop(columns=["SalePrice", "Id"])

numeric = X.select_dtypes(include="number").columns
categorical = X.select_dtypes(include="object").columns

preprocessor = make_column_transformer(
    (make_pipeline(SimpleImputer(strategy="median"),
                   StandardScaler()), numeric),
    (make_pipeline(SimpleImputer(strategy="most_frequent"),
                   OneHotEncoder(handle_unknown="ignore")), categorical)
)

lasso = make_pipeline(
    preprocessor,
    Lasso(max_iter=100000, tol=0.001)
)
```

```python
search = GridSearchCV(
    lasso,
    {"lasso__alpha": np.logspace(0, 4, 30)},
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

search.fit(X, y)

print("Best alpha:", search.best_params_)
print("Best CV RMSE:", -search.best_score_)
```

```python
model = search.best_estimator_

feature_names = model.named_steps[
    "columntransformer"
].get_feature_names_out()

coefficients = model.named_steps["lasso"].coef_

selected_features = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

selected_features = selected_features[
    selected_features["Coefficient"] != 0
]

selected_features
```

```python
selected_features['Abs_Coefficient'] = selected_features['Coefficient'].abs()
top_features = selected_features.sort_values(by='Abs_Coefficient', ascending=False)
top_features[['Feature', 'Coefficient']].head(15)
```

### Predictive Modeling

```python

# 1. Redefine the preprocessor just in case it was lost from memory
numeric = X.select_dtypes(include="number").columns
categorical = X.select_dtypes(include="object").columns

preprocessor = make_column_transformer(
    (make_pipeline(SimpleImputer(strategy="median"), StandardScaler()), numeric),
    (make_pipeline(SimpleImputer(strategy="most_frequent"), OneHotEncoder(handle_unknown="ignore")), categorical)
)

# 2. Create a new pipeline using the preprocessor
rf_pipeline = make_pipeline(
    preprocessor,
    RandomForestRegressor(n_estimators=100, random_state=42)
)

# 3. Evaluate with Cross-Validation to compare against Lasso
rf_scores = cross_val_score(
    rf_pipeline, 
    X, y, 
    cv=5, 
    scoring="neg_root_mean_squared_error", 
    n_jobs=-1
)

print(f"Random Forest CV RMSE: {-rf_scores.mean():.2f}")
```

```python
y_log = np.log1p(y)

# 1. Evaluate Log-RMSE (Target: <= 0.145 for tree-based, <= 0.18 for Lasso/regression)
log_rmse = -cross_val_score(
    rf_pipeline, X, y_log, cv=5, scoring="neg_root_mean_squared_error", n_jobs=-1
).mean()

# 2. Evaluate R^2 (Required: > 0.1 for Progress Check)
r2 = cross_val_score(
    rf_pipeline, X, y_log, cv=5, scoring="r2", n_jobs=-1
).mean()

print(f"CV Log-RMSE: {log_rmse:.4f}")
print(f"CV R^2:      {r2:.4f}")
```

```python
# 1. Fit on all training data (log-transformed target)
rf_pipeline.fit(X, y_log)

# 2. Load test.csv and predict in log-dollars
test = pd.read_csv('/Users/ryleighbergmann/House_Prices/Cleaned Data/cleaned_test.csv')
log_preds = rf_pipeline.predict(test.drop(columns=["Id"]))

# 3. Format exactly as Kaggle requires: Id, SalePrice (in regular dollars, no index column)
submission = pd.DataFrame({
    "Id": test["Id"],
    "SalePrice": np.expm1(log_preds)
})
submission.to_csv("submission.csv", index=False)

# 4. Verify format
print(submission.shape)   # Must be (1459, 2)
print(submission.head(3)) # Must show Id 1461+ and dollar prices
```
